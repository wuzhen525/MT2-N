```python
from sentence_transformers import SentenceTransformer
import array

os.environ["HTTP_PROXY"] = "http://127.0.0.1:7890"
os.environ["HTTPS_PROXY"] = "http://127.0.0.1:7890"
```

```python
df = pd.read_csv('label_desc.csv')
print(df.isnull().sum())
df.head()
```

```python
sentences = []
for i, row in df.iterrows():
    sen = str(row['product_name']) + ' ' + row['first-level category'] + ' ' + row['secondary category'] + ' ' \
    + row['characteristics'] + ' ' + row['function and purpose'] + ' ' + row['applicable population'] + ' ' \
    + row['related items']

    sentences.append(sen)

sentences[:5]
```

```python
model = SentenceTransformer('all-MiniLM-L6-v2')
sentence_embeddings = model.encode(sentences)
sentence_embeddings[:5]
```

```python
np.save('text_feat_au.npy', sentence_embeddings)
text_feat = np.load('text_feat_au.npy', allow_pickle=True)
print(text_feat.shape)
text_feat[:5]
```
